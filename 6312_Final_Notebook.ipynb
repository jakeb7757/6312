```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import kagglehub
from matplotlib.ticker import FuncFormatter

sns.set_theme(style='ticks')

# Download latest version
path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")
path = '/kaggle/input/ibm-hr-analytics-attrition-dataset/WA_Fn-UseC_-HR-Employee-Attrition.csv'

df = pd.read_csv(path)

print(df.head())
```

# EDA

```python
print('Missing Values per Column')
print(df.isnull().sum())
print('-------------')
print('DataFrame Shape')
print(df.shape)
print('-------------')
print('Data Types:')
print(df.dtypes)
```

```python
print("Summary Statistics:")
display(df.describe(include="all"))
```

```python
sns.histplot(
    data=df,
    x='YearsAtCompany',
    hue='Attrition',
    bins=12
)

plt.title('Distribution of Years at Company')
plt.ylabel('Count')
plt.xlabel('Years at Company')

plt.tight_layout()
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Attrition',
    y='DistanceFromHome'
)

plt.title('Distance From Home by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Distance From Home (mi)')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Attrition',
    y='Age'
)

plt.title('Age by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age (Years)')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome'
)

plt.title('Monthly Income by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')

plt.show()
```

```python
df['AttritionNum'] = df['Attrition'].map({'Yes': 1, 'No': 0})

jobRole = df.groupby('JobRole')['AttritionNum'].mean().reset_index().sort_values('AttritionNum', ascending=False)

print(jobRole)

plt.figure(figsize=(12,5))

sns.barplot(
    data=jobRole,
    x='AttritionNum',
    y='JobRole',
    orient='h'
)

sns.despine()

plt.title('Attrition Rate by Job Role')
plt.xlabel('Attrition Rate')
plt.ylabel('Job Role')

plt.tight_layout()
plt.show()
```

```python
overTime = df.groupby('OverTime')['AttritionNum'].mean().reset_index().sort_values('AttritionNum', ascending=False)

print(overTime)

plt.figure(figsize=(12,5))

sns.barplot(
    data=overTime,
    x='AttritionNum',
    y='OverTime',
    orient='h'
)

plt.title('Attrition Rate by Overtime')
plt.xlabel('Attrition Rate')
plt.ylabel('Overtime')

plt.tight_layout()
plt.show()
```

```python
df['OverTimeNum'] = df['OverTime'].map({'Yes': 1, 'No': 0})
overTimeRole = df.groupby('JobRole')['OverTimeNum'].mean().reset_index().sort_values('OverTimeNum', ascending=False)

print(overTimeRole)

plt.figure(figsize=(12,5))

sns.barplot(
    data=overTimeRole,
    x='OverTimeNum',
    y='JobRole',
    orient='h'
)

plt.title('Overtime Rate by Role')
plt.xlabel('Overtime Rate')
plt.ylabel('Role')

plt.tight_layout()
plt.show()
```

```python
satisfaction = df.groupby('JobRole')['JobSatisfaction'].mean().reset_index().sort_values('JobSatisfaction', ascending=False)

print(satisfaction)

plt.figure(figsize=(12,5))

sns.barplot(
    data=satisfaction,
    x='JobSatisfaction',
    y='JobRole',
    orient='h'
)

plt.title('Job Satisfaction by Role')
plt.xlabel('Job Satisfaction')
plt.ylabel('Role')

plt.tight_layout()
plt.show()
```

```python
satisfactionAttrition = df.groupby('JobSatisfaction')['AttritionNum'].mean().reset_index().sort_values('AttritionNum', ascending=False)

plt.figure(figsize=(12,5))

sns.barplot(
    data=satisfactionAttrition,
    x='AttritionNum',
    y='JobSatisfaction',
    orient='h'
)

plt.title('Attrition Rate by Satisfaction')
plt.xlabel('Attrition Rate')
plt.ylabel('Job Satisfaction')

plt.tight_layout()
plt.show()
```

```python
timeineachrole = df.groupby('JobRole')['YearsInCurrentRole'].mean().reset_index().sort_values('YearsInCurrentRole', ascending=False)

print(timeineachrole)

plt.figure(figsize=(12,5))

sns.barplot(
    data=timeineachrole,
    x='YearsInCurrentRole',
    y='JobRole',
    orient='h'
)

plt.title('Years in Current Role by Role')
plt.xlabel('Years in Current Role')
plt.ylabel('Role')

plt.tight_layout()
plt.show()
```

```python
wlbAtrrition = df.groupby('WorkLifeBalance')['AttritionNum'].mean().reset_index().sort_values('AttritionNum', ascending=False)

print(wlbAtrrition)

plt.figure(figsize=(12,5))

sns.barplot(
    data=wlbAtrrition,
    x='AttritionNum',
    y='WorkLifeBalance',
    orient='h'
)

plt.title('Attrition Rate by Work Life Balance')
plt.xlabel('Attrition Rate')
plt.ylabel('Work Life Balance (4 Being Best)')

plt.tight_layout()
plt.show()
```

```python
roleOvertime = df.groupby(['JobRole', 'OverTime'])['AttritionNum'].mean().reset_index().sort_values('AttritionNum', ascending=False)

print(roleOvertime)

plt.figure(figsize=(12,8))

sns.barplot(
    data=roleOvertime,
    x='AttritionNum',
    y='JobRole',
    hue='OverTime',
    orient='h'
)

plt.title('Attrition Rate by Role Split by Overtime')
plt.xlabel('Attrition Rate')
plt.ylabel('Role')

plt.tight_layout()
plt.show()
```

# Polished Charts

```python
colors = ['red' if role == 'Sales Representative' else 'steelblue'
          for role in jobRole['JobRole']]

fig, ax = plt.subplots(figsize=(12,5))

sns.barplot(
    data=jobRole,
    x='AttritionNum',
    y='JobRole',
    orient='h',
    ax=ax,
    palette=colors,
)

ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x:.0%}'))
ax.set_title('Sales Reps are Leaving at More than Double the Company Average')
ax.set_xlabel('Attrition Rate')
ax.set_ylabel('Job Role')
plt.axvline(x=0.16, color='red', linestyle='--', linewidth=1, label='Company Average (16%)')
plt.legend()

sns.despine(ax=ax)

plt.tight_layout()
plt.savefig('chart1_job_role.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,10))

sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome',
    ax=ax
)

sns.despine()

ax.set_title('Employees Leaving the Company Earn Less Income')
ax.set_xlabel('Attrition')
ax.set_ylabel('Monthly Income ($)')

plt.tight_layout()
plt.savefig('chart2_income.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12,5))

sns.barplot(
    data=overTime,
    x='AttritionNum',
    y='OverTime',
    orient='h',
    ax=ax
)

sns.despine()
ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x:.0%}'))
ax.set_title('Employees Working Overtime Leave 3x More Often than Employees Not Working Overtime')
ax.set_xlabel('Attrition Rate')
ax.set_ylabel('Overtime')
plt.axvline(x=0.16, color='red', linestyle='--', linewidth=1, label='Company Average (16%)')
plt.legend()

plt.tight_layout()
plt.savefig('chart3_overtime2.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12,8))

sns.barplot(
    data=roleOvertime,
    x='AttritionNum',
    y='JobRole',
    hue='OverTime',
    orient='h',
    ax=ax
)

sns.despine()
ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x:.0%}'))
ax.set_title('Overtime Doubles Attrition Rate Within Roles ')
ax.set_xlabel('Attrition Rate')
ax.set_ylabel('Role')
plt.axvline(x=0.16, color='red', linestyle='--', linewidth=1, label='Company Average (16%)')
plt.legend()

plt.tight_layout()
plt.savefig('chart4_overtime_split.png', dpi=300, bbox_inches='tight')
plt.show()
```
